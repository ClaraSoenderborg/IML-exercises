### Focus b: Matrices, matrix operations and their applications
In week 3's exercises we practiced using matrix operations. Matrices are 2D arrays of scalar values, and are central for machine learning. We talk about a matrix' shape or dimensions by its number of rows and number of columns. In this exercise we do matrix multiplication, elementwise multiplication and elementwise difference. 

Matrix operations
1. elementwise difference - subtraction
2. addition can be done the same way, if two matrices are the same size
4. matrix multiplication is a central operation in machine learning. When multiplying two matrices A and B, the output is a new matrix with A-rows and B-columns. Not a symmetric operation, and A's columns must be the same as B's rows. Each element in the output matrix is an inner product of row in A and column in B. 
6. transposing a matrix turns its rows into columns and vice versa. So the first column becomes the first row in the transposed matrix. 
7. matrix inverse. Go to W04, 01-linear_algebra. A square matrix has an inverse if its determinant is not 0, which means that the columns of the matrix span a space. A matrix multiplied with its inverse equals the identity matrix. 

In week 4s exercises, we use the matrix inverse to solve linear systems of equations. 
In week 4s tutorial, we use matrix multiplication to do linear transformations.
