# Q4 - mandatory 1

### focus d: vector spaces, basis and independence

In this exercise, we worked with vector spaces.

A vector space is a set of vectors corresponding to all linear combinations of the basis vectors, which is a set of linearly independent vectors. In our normal 2d coordinate system, we have the standard basis $\begin{bmatrix}1 & 0 \\ 0 & 1\end{bmatrix}$, which is the identity matrix. This is the same for 3d and so on. With these two column vectors in the standard basis, we can get any 2d vector by linear combinations. To work in a different vector space, we need a basis formed of different vectors, that are still independent from each other. We can have a 2d subspace in a 3d space by only having two independent 3d vectors. 

In the first question, we have a 3d space with a plane and a line. This plane can be seen as a 2d subspace and the line is a 1d subspace. If they both should be contained in a 2d subspace, they should be on top of each other, with same slope and bias and so on. 

In the second question, we're asked if systems of equations have a solution. We solve these systems of solution by inversing the matrix. We can only inverse if the column vectors are linearly independent, so they span an nd space where n is also the size of the matrix. This is why we usually find the determinant of the matrix, as if this is 0 means that some of the column vectors have the same direction and are linearly dependent. Only a have a linearly independent matrix. We can also say that the matrix' rank is equal its dimensions

In the third question, we have that two vectors are linearly independent, such that they span a space. Then a third vector is linearly dependent on these two, meaning that it exists in the two vectors span. Draw this.
