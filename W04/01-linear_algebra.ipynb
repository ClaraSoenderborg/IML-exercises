# Q2: Exercise week 4 (linear transformations)
### Introduction:
Transformation is changing data within same domain or to another domain. You have some transformation matrix T, input vector x and output vector y. $y = Tx$
For a transformation to be linear, we have some properties that must hold. For vectors u and v, $T(u+v) = T(u) + T(v)$. And scaling the transformation with some constant c, must be the same as scaling the vector with c and then transforming it with T. Visually, we can thnk of a transformation as changing the entire space, like the coordinate system. For the transformation to be linear, the origin must stay in the same place, and lines must remain lines. Nothing should start curving. 
We can think of the columns of T to be basis vectors for this transformed space. In the standard basis, we have basis vectors $[1,0]$ and $[0,1]$. Applying a transformation to a data point is therefore like a linear combination. 

This week's exercises are about using matrix inverses to solve linear systems of equations, and using matrix inverses to estimate model parameters in single variate polynomials. Both exercises uses linear transformations to solve theses problems. 

**Focus a. linear transformations:**

With linear transformations, you can augment the data. We can think of it as changing the underlying coordinate system, not the actual data. 

Show scaling -> x value is scaled by 2, y value is scaled by 3. In higher dimension, add scaling value for each axis. 

Show shearing -> horizontal or vertical

Show rotation -> around origin or around point

3d rotation -> around axis, for higher dimensions 

Remember, for linear transformations, the origin is fixed and all lines stay as lines, as well as staying parallel and evenly spaced. If we want to translate the data, shifting it by some bias, that is not linear, but affine, as we move the origin. To express translations in a single matrix multiplication, we use homogeneous coordinates, which adds another dimension to be able to add this bias. 

There are more non-linear transformations, where the lines curve and so on. 


**Focus b. learning with matrix inverses:**

If we have a dataset, and we want to train a regression model, we need to learn the coefficients. 
For task 5, we solve linear systems of equations using linear algebra. We structure these equations into an inner product $Ax = b$. If we want to isolate x, we can multiply the inverse of A on both sides, such that $x = A^{-1}b$. This works because a matrix multiplied with its own inverse is the identity matrix. Multiplying a vector or matrix with the identity matrix does not change the input. Inverses only exist for square matrices, and the determinant of the matrix must be non-zero. The columns of the matrix must span a space, and not lie in the same direction (the matrix is singular). See for example the last system of equations. 

We do the same when learning model parameters for a polynomial regression model. We form a design matrix, similar to solving the linear equations, containing the input data. Because we also need a bias in our model, we use homogeneous coordinates in our design matrix. Because we need our design matrix to be square, for it to have an inverse, we need a certain amount of data points depending on the complexity of the model. An affine model with the underlying function $y=ax+b$ need 2 points, because we have 2 unknowns. A 2nd order polynomial needs 3 data points and so on. If we had a linear model $y=ax$, then we know that the line hits the origin (0,0). For a 2nd order polynomial with a bias, the rows of the design matrix would be $[x, x^2, 1]$, with 3 columns, and it therefore also needs 3 rows of data points. 

**Focus c. affine transformation, homogeneous coordinates, composition of linear transformation**

Affine transformations are a combination of linear transformation with a translation, meaning that we shift all points in a certain direction. For some linear transformation, we have use some matrix T, where if the data is 2d, then T is 2x2. But how do we do translations with this matrix? How do we move the input based on the origin? We want to add some bias to the data, not scale it. We use homogeneous coordinates for this, which we can also view as projective coordinates. By adding a dimension, we can also shift the data with a bias. 

For a simple translation,
$\begin{bmatrix}1 & 0 & dx \\ 0 & 1 & dy \\ 0 & 0 & 1\end{bmatrix} \begin{bmatrix}x \\ y \\ 1\end{bmatrix}$

If we have several transformations, such as an affine transformation that rotates and translates, we can multiply these matrices to combine them = composition of linear transformation

## Linear equations
Matrices can represent systems of linear equations

$$
Ax=b
$$
where $A$ is the coefficient matrix, $x$ the vector of unknowns, and $b$ is the vector of the dependent variables.
A solution can be found using

$$
\begin{align*}
A^{-1}Ax&=A^{-1}b\\
x &= A^{-1}b.
\end{align*}
$$

For an $N\times D$ matrix $A$ and a $D\times K$ matrix $B$, the 
matrix multiplication (or matrix product) is an $N\times K$ matrix $R$. Elements $R_{ij}$ of $R$ can be calculated 
using the following formula

$$
R_{ij} = \sum_{d=1}^D A_{id}B_{dj}.
$$
In other words, it is the dot product of the $i$'th row vector of $A$ and the $j$'th column vector of $B$.
