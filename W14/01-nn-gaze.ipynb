### Focus b: NN training (gradients, chain rule, forward and back propagation)
How do we learn these model weights? Many more than simple linear model. Non-linear cannot use LLS. 
LLS is a linear optimization method which minimizes sum of squared differences. MSE can be seen as the loss function, taking the weights as input and outputting the MSE. For a simple 2d linear regression model, the loss function could look like a 2d sheet with a dip somewhere inside a 3d space. 

Intuition is that we want to find global minimum, and we can imagine standing somewhere on the line and walking downhill until we can't. Difficult with multiple local minimums and multiple dimensions. 

A gradient vector contains all partial derivatives of each parameter. The gradient of the loss function for some input will point in the direction of the greatest increase. We therefore have to walk in the negative gradients direction. This is done in small steps, learning rate for some iterations (epochs). Training loss curve shows relationship between two hyperparameters. 

To actually do gradient descent, we use back propagation. Give some random weights (like starting at random point for gradient descent) and do forward propagation. Then calculate loss with training data to get the loss. Work backward through neural network, adjusting weights. Looking at a single weight, what is the deritivate of the loss function with respect to that weight? Chain rule, as this function is composed of multiple functions. 

### Focus c: Training and evaluation (over and underfitting)
In this week exercise, we used our gaze data for a gaze estimation model. Earlier we made an affine regression model to predict, and while it did okay, we had a bias error that suggested the model was underfitted. 

We try to use a non-linear model, by passing output through relu activation function. When training, we have to choose hyper parameters such as learning rate and epoch. Learning rate is size of steps taken during gradient descent, a scale to the negated gradient. Epochs are like iterations of training. If we have a small learning rate, we usually need a large epoch to reach a minimum in the loss function. 

To discover how the hyperparameters work, we create several models, either single layer or multi layer with different learning rates and epochs. 

We have split data into train, validation and test data. We can compare training error with the validation error to understand the generalizability of our model, and whether its over or underfitted. To evaluate the overall performance of the model with our test data, we look at the mean squared error. Models with several layers add more complexity, and could become overfitted. We should look for cases where the training error decreases and the validation error increases. This is not the case for any of our models. Some simply have a high bias, which suggests underfitting due to our hyper parameters. 

### Focus d: Model architecture, MLP vs CNN
What I have shown so far are MLP that are fully connected. All neurons of a layer are scaled by some weights and summed in all neurons of next layer. This gives us a lot of weights to optimise when training, and can be very complex. Too complex = overfitting. See relu models, where output is passed through activation function in the forward function. 

If we have large complex image as input, and maybe we want to classify that. If each pixel is an input neuron, we have a lot of weights. 

Convolutional neural networks are different, as we apply filters through the edges of the network. If the input is some image, then each pixel in window of the image is connected to some output pixel of that filter. It is no longer fully connected. Example with 4x4 image and 3x3 gaussian blur filter, producing 3x3 output image. 

After convolution step, apply max pooling or other pooling to get smaller output. This can then be flattened and given to a fully connected MLP. 


### Focus e: Loss function, model complexity, cross validation
How do we learn these model weights? Many more than simple linear model. Non-linear cannot use LLS. 
LLS is a linear optimization method which minimizes sum of squared differences. MSE can be seen as the loss function, taking the weights as input and outputting the MSE. For a simple 2d linear regression model, the loss function could look like a 2d sheet with a dip somewhere inside a 3d space. With a lot of weights, and a non-linear relationship between weights, have to do gradient descent. Multiple local minimums.

Loss function and its landscape become more complex the more layered and more connected the model is, we get more weights. Should be as complex as needed for the problem. Occams razor. See relu models exercise. 

To decide the models complexity, hyperparameters, train each model and compare. Even better if we performed k-fold cross validation, so each model is trained multiple times with different data and then tested on different data.

```python
class LinearModel(nn.Module):
    """
    Args:
        input_dim (int): Number of input features.
        output_dim (int): Number of output features.

    Methods:
        forward(x): Passes the input through the linear layer.
    """
    def __init__(self, input_dim, output_dim):
        super(LinearModel, self).__init__()
        self.linear1 = nn.Linear(input_dim, output_dim)
 
    def forward(self, x):
        """Args:
        x (Tensor): Input tensor.
        Returns:
            Tensor: Output tensor after applying the linear transformation.
        """
        x = self.linear1(x)
        return x
    
class MSELoss(nn.Module):
    def __init__(self, reduction='mean'):
        super(MSELoss, self).__init__()
        self.reduction = reduction

    def forward(self, input, target):
        squared_diff = (input - target) ** 2
        if self.reduction == 'mean':
            return squared_diff.mean()
        elif self.reduction == 'sum':
            return squared_diff.sum()
        else:
            raise ValueError("Invalid reduction type. Use 'mean' or 'sum'.")

    
def train_model(model, criterion, optimizer, X_train, Y_train, X_val=None, Y_val=None, num_epochs=100):
    """
    Args:
        model (nn.Module): The neural network model to train.
        criterion (nn.Module): The loss function to minimize.
        optimizer (torch.optim.Optimizer): Optimizer for updating model parameters.
        X_train (Nx2 Tensor): Training input data.
        Y_train (Nx2 Tensor): Training target data.
        X_val (Nx2 Tensor, optional): Validation input data. Defaults to None.
        Y_val (Nx2 Tensor, optional): Validation target data. Defaults to None.
        num_epochs (int): Number of training epochs.

    Returns:
        list: Loss values for each epoch (training).
        list: Loss values for each epoch (validation).
        float: Training time.
    """
    start_time = time.time()
    train_losses = []
    val_losses = []
    model_params = []

    for epoch in range(num_epochs):
        model.train()
        optimizer.zero_grad()
        outputs = model(X_train)
        loss = criterion(outputs, Y_train)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 0.5)
        optimizer.step()
        train_losses.append(loss.item())


        with torch.no_grad():
            val_outputs = model(X_val)
            model_params.append(model.parameters())
            val_loss = criterion(val_outputs, Y_val)
            val_losses.append(val_loss.item())

    end_time = time.time()
    training_time = end_time - start_time

    return train_losses, val_losses, training_time



def test_model(model, X_test, Y_test):
    """
    Evaluates a trained model on test data.

    Args:
        model (nn.Module): The trained neural network model.
        X_test (Tensor): Test input data.
        Y_test (Tensor): Test target data.

    Returns:
        float: Mean squared error (MSE) over the test set.
        np.ndarray: Predicted values as a numpy array.
        np.ndarray: True values as a numpy array.
        np.ndarray: Absolute errors for x and y coordinates.
    """
    model.eval()
    with torch.no_grad():
        test_output = model(X_test)
        mse = mean_squared_error(Y_test.cpu().numpy(), test_output.cpu().numpy())
        predictions = test_output.cpu().numpy()
        true_values = Y_test.cpu().numpy()
        errors = np.abs(true_values - predictions)

    return mse, predictions, true_values, errors



# Set parameters
input_dim = 2
output_dim = 2
learning_rate = 0.1
epoch = 20000

# Generate training data
X_train_tensor = torch.tensor(X_train, dtype=torch.float32)
Y_train_tensor = torch.tensor(Y_train, dtype=torch.float32)  
X_test_tensor = torch.tensor(X_test, dtype=torch.float32)
Y_test_tensor = torch.tensor(Y_test, dtype=torch.float32)  
X_val_tensor = torch.tensor(X_val, dtype=torch.float32)
Y_val_tensor = torch.tensor(Y_val, dtype=torch.float32)   

model = LinearModel(input_dim, output_dim)
criterion = MSELoss()
optimizer = optim.SGD(model.parameters(), lr=learning_rate)

# Train the model
losses, val_losses, training_time = train_model(model, criterion, optimizer, X_train_tensor, Y_train_tensor, X_val_tensor, Y_val_tensor, num_epochs=epoch)

# Test the model
mse, Y_pred_lm, true_values, errors_nn = test_model(model, X_test_tensor, Y_test_tensor)
print(f'Average MSE: {mse}')

# Visualize results
nn_util.plot_results(
        Y_test_tensor,
        Y_pred_lm,
        errors_nn,
        losses,
        val_losses,
        model_name='NN',
        training_time=training_time
    )
```

```python
class DataScaler:
    def __init__(self):
        self.min = None
        self.max = None

    def normalize(self, data):
        self.min = np.min(data)
        self.max = np.max(data)
        normalized_data = (data - self.min) / (self.max - self.min)
        return normalized_data

    def denormalize(self, normalized_data):
        data = normalized_data * (self.max - self.min) + self.min
        return data
    

scaler_input = DataScaler()
scaler_target = DataScaler()

X_train_norm = scaler_input.normalize(X_train)
X_test_norm = scaler_input.normalize(X_test)
X_val_norm = scaler_input.normalize(X_val)

Y_train_norm = scaler_target.normalize(Y_train)
Y_test_norm = scaler_target.normalize(Y_test)
Y_val_norm = scaler_target.normalize(Y_val)

# Generate training data
X_train_tensor = torch.tensor(X_train_norm, dtype=torch.float32)
Y_train_tensor = torch.tensor(Y_train_norm, dtype=torch.float32)  
X_test_tensor = torch.tensor(X_test_norm, dtype=torch.float32)
Y_test_tensor = torch.tensor(Y_test_norm, dtype=torch.float32)  
X_val_tensor = torch.tensor(X_val_norm, dtype=torch.float32)
Y_val_tensor = torch.tensor(Y_val_norm, dtype=torch.float32)   

# Set hyperparameters
input_dim = 2
output_dim = 2
learning_rate = [0.0001, 0.1, 0.5]
epoch = [500, 2000, 10000]
criterion = MSELoss()

# Containers gaze data
models_dict = {}
losses_train_dict = {}
losses_val_dict = {}
training_time_dict = {}
pred_norm_dict = {}
errors_norm_dict = {}
mse_norm_dict= {}



#Train the models
for i in learning_rate:
    for j in epoch:
        model_name = f'LinearModel_lr{i}_ep{j}'
        model = LinearModel(input_dim, output_dim)
        criterion = MSELoss()
        optimizer = optim.SGD(model.parameters(), lr=i)

        # Train the model
        losses, val_losses, training_time = train_model(model, criterion, optimizer, X_train_tensor, Y_train_tensor, X_val_tensor, Y_val_tensor, num_epochs=j)

        # Test the model
        mse, Y_pred_lm, true_values, errors_nn = test_model(model, X_test_tensor, Y_test_tensor)
        
        
        models_dict[model_name] = model
        losses_train_dict[model_name] = losses
        losses_val_dict[model_name] = val_losses
        training_time_dict[model_name] = training_time
        pred_norm_dict[model_name] = Y_pred_lm
        errors_norm_dict[model_name] = errors_nn
        mse_norm_dict[model_name] = mse



nn_util.plot_mse_bar(mse_norm_dict)
nn_util.plot_results_collected(Y_test_tensor, pred_norm_dict, errors_norm_dict, models_dict, losses_train_dict, losses_val_dict, training_time_dict)

```

```python
class NNRelu(nn.Module):
    def __init__(self, input_dim, output_dim):
        super(NNRelu, self).__init__()
        self.linear1 = nn.Linear(input_dim, output_dim)
        self.relu = nn.ReLU()
 
    def forward(self, x):
        x = self.linear1(x)
        x = self.relu(x)
        return x
 
class NNRelu_exp(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super(NNRelu_exp, self).__init__()
        self.linear1 = nn.Linear(input_dim, hidden_dim)
        self.relu1 = nn.ReLU()
        self.linear2 = nn.Linear(hidden_dim, hidden_dim)
        self.relu2 = nn.ReLU()
        self.linear3 = nn.Linear(hidden_dim, output_dim)
 
    def forward(self, x):
        x = self.linear1(x)
        x = self.relu1(x)
        x = self.linear2(x)
        x = self.relu2(x)
        x = self.linear3(x)
        return x
```
