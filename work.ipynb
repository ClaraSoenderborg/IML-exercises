```python
import numpy as np

X = np.array([
    [1, 1, 1],
    [1, 1, 2],
    [1, 2, 2],
    [1, 3, 4],
    [1, 5, 3],
    [1, 6, 2]
])

y = np.array([1, 1, 1, -1, -1, -1]).T

w = np.linalg.lstsq(X, y)[0]

print(w)

import matplotlib.pyplot as plt  

x_values = np.linspace(0, 10, 100)


plt.scatter(X[:,1], X[:,2])
plt.plot(x_values, -(w[0]/w[2])-(w[1]/w[2])*x_values)
```
